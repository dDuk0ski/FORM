# Workout ML Testing
============================================================================================================================

The goal is to implement a machine learning algorithm that allows for adaption to each individual person as opposed to training a large model on general data the goal is to take the inputs of things like sleep, previous workouts, amount eaten, level of fatigue and other values as input and then is able to give you an optimized workout plan based on past experiences that have worked for you as an individual with it being able to suggest certain exercises, amount of sets, rep ranges, rest between sets, weight for each set, etc. 

The current idea is to use a method known as Bayesian Updating or Bayesian Inference which essential has a belief that it uses before obtaining new evidence and updating its beliefs accordingly 

This will need a few different steps this model will need to take as since the user is starting with no knowledge of the user or what works best for them. To start we need to first train a baseline model that is able to produce a tangible realistic workout plan that can then be updated through the new knowledge of the user obtained from subsequent workouts and other dailt metrics.

My current proposal for a pipeline is as such:
1. Create or train an algorithm or method that is capable of creating a realistic workout based around specific muscle groups or movements and how to order them and handle things like weight, number of sets, number of reps, etc
2. Next all the other user information being taken from elsewhere within the application will be taken and assembled to be used as training data for the Bayesian Inference 
3. (OPTIONAL) While not required giving the Bayesian Inference some generic beliefs could accelerate the progress meaning a pretrained baseline on some either generated or available data online could be used to decrease the amount of time before meaningful workout generation
4. Now is when the Bayesian Inference begins working where each new user workout logged along side the next day to see how effective the workout was and update beliefs (beliefs should also be able to take a retroactive look maybe based on weeks or months later to updated the long term effect of the scheduled training vs the immediate consequences)
